## Helper Functions

Utility functions for consistent figure styling.

```python
def add_panel_label(ax, label, x=-0.1, y=1.05):
    """Add panel label (A, B, C, etc.) to subplot"""
    ax.text(x, y, label, transform=ax.transAxes,
            fontsize=14, fontweight='bold', va='top', ha='right')

def add_significance_bar(ax, x1, x2, y, h, p_value):
    """Add significance bar and asterisks to plot"""
    # Draw horizontal line
    ax.plot([x1, x1, x2, x2], [y, y+h, y+h, y], lw=1.5, c='black')
    # Add asterisks based on p-value
    if p_value < 0.001:
        text = '***'
    elif p_value < 0.01:
        text = '**'
    elif p_value < 0.05:
        text = '*'
    else:
        text = 'n.s.'
    ax.text((x1+x2)*0.5, y+h, text, ha='center', va='bottom', fontsize=12)

def format_axis(ax):
    """Apply consistent axis formatting"""
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)
    ax.tick_params(labelsize=10)
    ax.xaxis.label.set_size(11)
    ax.yaxis.label.set_size(11)

print('Helper functions defined')
```
